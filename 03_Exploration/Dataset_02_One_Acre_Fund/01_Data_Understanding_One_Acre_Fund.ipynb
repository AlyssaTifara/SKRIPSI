## 4. Periode Waktu, Cakupan Geografis, dan Unit Observasi

### 4.1 Periode Pengumpulan Data
Berdasarkan kolom `year`, survei MEL (*Monitoring, Evaluation, and Learning*) One Acre Fund mencakup periode tahun 2016 hingga 2022.


```python
year_dist = df_oaf['year'].value_counts().sort_index().reset_index()
year_dist.columns = ['Tahun', 'Jumlah Observasi']
year_dist['Persentase (%)'] = (year_dist['Jumlah Observasi'] / len(df_oaf) * 100).round(2)
year_dist
```

*Catatan Observasi Periode (Fakta Data vs Batasan Interpretasi):*
- Survei terbesar terkumpul pada rentang 2018–2020 (total 11.127 observasi atau 75,32%).
- **Tahun 2021 tidak tercatat** dalam berkas survei yang tersedia (hanya mencakup tahun 2016, 2017, 2018, 2019, 2020, dan 2022). Ketiadaan data tahun 2021 dicatat murni sebagai **fakta struktur dataset**. Tidak ada dokumentasi resmi dalam repositori yang menjelaskan penyebab ketiadaan survei pada tahun tersebut, sehingga penyebabnya tidak diasumsikan.
- Tahun 2022 hanya memuat 115 observasi yang seluruhnya berasal dari Nigeria.


### 4.2 Negara dan Cakupan Geografis
Dataset mencakup 7 negara di kawasan Afrika Sub-Sahara (*Sub-Saharan Africa*): Kenya, Tanzania, Rwanda, Zambia, Burundi, Nigeria, dan Uganda.


```python
country_dist = df_oaf.groupby(['country', 'iso']).size().reset_index(name='Jumlah Observasi')
country_dist['Persentase (%)'] = (country_dist['Jumlah Observasi'] / len(df_oaf) * 100).round(2)
country_dist.sort_values(by='Jumlah Observasi', ascending=False)
```

```python
# Tabulasi Silang Negara vs Tahun Survei
ct_country_year = pd.crosstab(df_oaf['country'], df_oaf['year'], margins=True, margins_name="Total")
ct_country_year
```

### 4.3 Musim Tanam (Season)
Variabel `season` mencatat dua siklus musim tanam utama: *first season* (musim tanam utama / *long rains*) dan *second season* (musim tanam kedua / *short rains*).


```python
season_ct = pd.crosstab(df_oaf['country'], df_oaf['season'], margins=True, margins_name="Total")
season_ct
```

*Fakta Musim Tanam:*
- Sebanyak 13.882 observasi (93,97%) berasal dari *first season*.
- Seluruh 891 observasi *second season* berasal eksklusif dari Rwanda, yang secara agroklimat memiliki dua musim tanam bimodal yang terdefinisi jelas.


### 4.4 Analisis Unit Observasi (*Unit of Analysis*)
Apakah 1 baris merepresentasikan seorang petani, satu rumah tangga, atau satu plot lahan?

1. **Struktur Identitas:** Dataset tidak memiliki kolom `farmer_id` atau `household_id` eksplisit. Kolom identifikasi terdekat adalah `strat` (*survey stratification* e.g., `01_KEN7701S1`) dan `cz` (*crop zone*).
2. **Koordinat Geografis:** Terdapat 4.128 nilai unik `lat` dan 3.375 nilai unik `lon` dari 14.773 baris. Hal ini menunjukkan bahwa titik koordinat merupakan representasi sentroid kluster desa (*village/cluster GPS*) demi privasi petani kecil (*smallholder farmer privacy*), bukan GPS mikro per meter persegi tanaman.
3. **Kesimpulan Unit Observasi:**
   > **Satu baris data merepresentasikan satu plot budidaya jagung (*maize field/plot*) milik petani mitra One Acre Fund yang dievaluasi pada satu siklus musim panen tertentu**, lengkap dengan karakteristik biofisik lingkungan, praktik input agronomi yang diaplikasikan, serta hasil panen riil (*crop-cut / harvest survey*) yang diukur dalam kilogram per hektar (`yield_kg_ha`).

## 6. Verifikasi Makna: Fakta Terkonfirmasi vs Interpretasi Sementara

Dalam metodologi ilmiah, pembedaan antara fakta yang didukung bukti dokumentasi dan asumsi sementara wajib ditegaskan secara transparan:

### A. Fakta yang Terkonfirmasi (Confirmed Facts)
1. **Target `yield_kg_ha`:** Terkonfirmasi sebagai hasil panen biji jagung standar per hektar (bukan ton/ha, bukan pound/acre) dengan nilai rata-rata $2.969,06	ext{ kg/ha}$ ($pprox 2,97	ext{ ton/ha}$), sangat konsisten dengan benchmark produktivitas jagung petani kecil di Afrika Sub-Sahara oleh One Acre Fund dan FAO.
2. **Karakteristik Kimia Tanah (`soil_pH`, `soil_clay`, `soil_ECEC`, `soil_rzpawhc`):** Terkonfirmasi berasal dari gridded soil database (seperti AfSIS / SoilGrids) yang dipetakan pada kluster geografis survei.
3. **Parameter Cuaca/Iklim (`season_prec`, `avg_season_tavg`, `avg_season_gdd`, `avg_season_ai`):** Terkonfirmasi merupakan parameter akumulasi dan rata-rata musiman hasil ekstraksi satelit/reanalysis cuaca (CHIRPS/ERA5).
4. **Variabel Biner Stres Lapangan (`pest`, `disease`, `water_excess`, `fert_in_hole`):** Terkonfirmasi lengkap 100% tanpa missing value, merupakan catatan checklist enumerator survei saat wawancara/inspeksi lapang sepanjang musim tanam.

### B. Variabel Belum Terverifikasi / Interpretasi Sementara (Provisional Interpretation)
1. **Definisi dan Titik Acuan `plant_date_dev`:** Rentang nilai berada pada $-108$ s.d. $+97$ hari dengan median 0 hari. Secara leksikal diinterpretasikan sebagai deviasi tanggal tanam terhadap suatu referensi (misalnya tanggal tanam optimal regional atau rencana tanam), namun formula matematis dan titik acuan referensinya **belum memiliki dokumentasi resmi dalam repositori**.
2. **Definisi dan Skala `hyb_mat`:** Nilai berada pada angka diskrit 0.5 hingga 3.0 (0.5, 1.0, 1.5, 2.0, 2.5, 3.0; median 2.5). Diduga merupakan kode klasifikasi kematangan One Acre Fund (misal: early, medium, late maturing), namun **belum terverifikasi oleh buku kode (*codebook*) resmi**.
3. **Satuan `soil_orgC`:** Nilai berkisar antara 3,48 hingga 39,45 dengan median 13,88. Angka ini secara agronomis diinterpretasikan merepresentasikan $g/kg$ ($14	ext{ g/kg} = 1,4\%	ext{ C-organik}$), bukan persen langsung.
4. **Penyebab Missingness pada Pupuk Organik:** Variabel `manure` memiliki missingness 94,00% dan `compost` 39,29%. Apakah nilai missing berarti petani sama sekali tidak menggunakan pupuk kandang ataukah pertanyaan survei dilewati pada kuisioner tahun tertentu belum dapat dipastikan secara definitif.
