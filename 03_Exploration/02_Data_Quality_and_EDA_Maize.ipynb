# **EDA B — Irigasi**

Analisis kedua berfokus pada aspek irigasi. Dataset Maize memiliki informasi mengenai jenis irigasi (`Irrigation_Type`) dan beberapa variabel lingkungan yang dapat digunakan untuk melihat kondisi tanaman, salah satunya `Soil_Moisture`.

Analisis ini dilakukan untuk mengetahui bagaimana karakteristik data berbeda berdasarkan jenis irigasi yang digunakan.

Fokus awal diberikan pada tiga hal:

1. Distribusi jumlah observasi berdasarkan jenis irigasi.
2. Perbedaan kondisi kelembapan tanah berdasarkan jenis irigasi.
3. Perbedaan hasil panen dan lama siklus tanam berdasarkan jenis irigasi.

Analisis ini bertujuan untuk mencari **bukti awal** apakah aspek irigasi memiliki pola yang cukup menarik untuk diteliti lebih lanjut.

Perbedaan nilai antarjenis irigasi pada tahap EDA tidak langsung diartikan sebagai bukti bahwa suatu metode irigasi lebih efektif. Perbedaan tersebut masih perlu dianalisis lebih lanjut dengan mempertimbangkan jumlah data, karakteristik data, serta faktor lain yang mungkin memengaruhi hasil.

```python
irrigation_counts = maize['irrigation_type'].value_counts(
    dropna=False
)

display(irrigation_counts)
```

```python
plt.figure(figsize=(8, 5))

irrigation_counts.plot(kind='bar')

plt.xlabel('Irrigation Type')
plt.ylabel('Number of Observations')
plt.title('Distribusi Jenis Irigasi pada Maize')
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

## **Irigasi vs Soil Moisture**

Analisis ini membandingkan distribusi `Soil_Moisture` berdasarkan `Irrigation_Type`.

`Soil_Moisture` digunakan sebagai salah satu indikator kondisi kelembapan tanah, sedangkan `Irrigation_Type` digunakan untuk membedakan metode irigasi.

Boxplot digunakan karena dapat memperlihatkan median, persebaran data, serta kemungkinan adanya nilai yang berbeda jauh dari sebagian besar observasi.

Tabel statistik kemudian digunakan untuk melihat jumlah data (`count`), rata-rata (`mean`), median, dan standar deviasi pada masing-masing jenis irigasi.

Tujuan analisis bukan untuk membuktikan efektivitas suatu jenis irigasi secara kausal, tetapi untuk mengetahui apakah terdapat perbedaan pola kelembapan tanah antarjenis irigasi yang dapat menjadi dasar analisis lebih lanjut.

```python
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=maize,
    x='irrigation_type',
    y='soil_moisture_%'
)

plt.xlabel('Irrigation Type')
plt.ylabel('Soil Moisture')
plt.title('Soil Moisture Berdasarkan Jenis Irigasi')

plt.tight_layout()
plt.show()
```

```python
irrigation_moisture = (
    maize
    .groupby('irrigation_type')['soil_moisture_%']
    .agg(['count', 'mean', 'median', 'std'])
    .round(2)
)

display(irrigation_moisture)
```

## **Irigasi vs Yield**

Analisis berikutnya membandingkan hasil panen (`Yield_kg_per_hectare`) berdasarkan jenis irigasi.

Tujuan analisis adalah mengetahui apakah terdapat perbedaan distribusi hasil panen pada observasi dengan jenis irigasi yang berbeda.

Boxplot digunakan untuk melihat persebaran hasil panen, sedangkan tabel statistik digunakan untuk membandingkan jumlah observasi, rata-rata, median, dan standar deviasi.

Hasil analisis ini dapat memberikan indikasi awal mengenai apakah variabel irigasi memiliki keterkaitan dengan hasil panen pada dataset.

Namun, perbedaan hasil panen antarjenis irigasi belum dapat dianggap sebagai bukti bahwa jenis irigasi tertentu menyebabkan hasil panen lebih tinggi. Hal tersebut karena hasil panen dapat dipengaruhi oleh banyak variabel lain, seperti kondisi tanah, suhu, curah hujan, kelembapan, pemupukan, dan kondisi tanaman.

```python
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=maize,
    x='irrigation_type',
    y='yield_kg_per_hectare'
)

plt.xlabel('Irrigation Type')
plt.ylabel('Yield (kg/hectare)')
plt.title('Yield Maize Berdasarkan Jenis Irigasi')

plt.tight_layout()
plt.show()
```

```python
irrigation_yield = (
    maize
    .groupby('irrigation_type')['yield_kg_per_hectare']
    .agg(['count', 'mean', 'median', 'std'])
    .round(2)
)

display(irrigation_yield)
```
